```python
from pyspark.sql import functions as F
from pyspark.sql import types as T
```

```python
# method - 1  using strict schema definition

cus_data = [
    (1, "John", "Smith", "john.smith@example.com", "123-456-7890"),
    (2, "Jane", "Doe", "jane.doe@example.com", "987-654-3210"),
    (3, "Bob", "Johnson", "bob.johnson@example.com", "555-123-4567"),
    (4, "Alice", "Williams", "alice.williams@example.com", "555-987-6543"),
    (5, "Charlie", "Brown", "charlie.brown@example.com", "555-555-5555")
]

cus_schema = T.StructType([
    T.StructField("customer_id", T.IntegerType(), True),
    T.StructField("first_name", T.StringType(), True),
    T.StructField("last_name", T.StringType(), True),
    T.StructField("email_address", T.StringType(), True),
    T.StructField("phone_number", T.StringType(), True)
])

cus_df = spark.createDataFrame(data=cus_data, schema=cus_schema)
display(cus_df)
```

#### Group By 

```python
sales_df = (
    spark.table("samples.bakehouse.sales_transactions")
)

sales_df.display()
```

```python
%sql
select
customerID,
sum(quantity) as total_quantity,
count(*) as total_transactions
from samples.bakehouse.sales_transactions
group by customerID;
```

```python
result_df = (
    sales_df
    .groupBy("customerID")
    .count()
)
result_df.display()
```

```python
result_df = (
    sales_df
    .groupBy("customerID")
    .agg(
        F.sum("quantity").alias("total_quantity"),
        F.count("*").alias("total_transactions"),
        F.min("quantity").alias("min_quantity"),
        F.max("quantity").alias("max_quantity")
    )
)
result_df.display()
```

```python
%sql
select
    customerID,
    product,
    sum(quantity) as total_quantity
from samples.bakehouse.sales_transactions
group by customerID, product;
```

```python
%sql
select
    customerID,
    product,
    sum(quantity) as total_quantity
from samples.bakehouse.sales_transactions
group by customerID, product
having sum(quantity) > 50;
```

```python
# Find the total_quantity of each product ordered by each customer -- cus_id, product, total_quality

result_df = (
    sales_df
    .groupBy("customerID", "product")
    .agg(
        F.sum("quantity").alias("total_quantity")
    )
    .filter(F.col("total_quantity") > 50)
)
result_df.display()
```

```python
sales_df = (
    spark.table("samples.bakehouse.sales_transactions")
)

sales_df.display()
```

```python
# Find the products list ordered by each customer

result_df = (
    sales_df
    .groupBy("customerID")
    .agg(
        F.collect_list(F.col("product")).alias("product_list")
    )
)

result_df.display()
```

```python
# Find the unique products list ordered by each customer

result_df = (
    sales_df
    .groupBy("customerID")
    .agg(
        F.collect_set(F.col("product")).alias("product_list")
    )
)

result_df.display()
```

```python
result_df = sales_df.groupBy("customerID").agg(F.collect_set(F.col("product")).alias("product_list"))
result_df.display()
```
