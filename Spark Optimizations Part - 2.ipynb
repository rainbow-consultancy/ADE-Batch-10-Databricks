```python
from pyspark.sql import functions as F
```

```python
# This file size is 1 GB
new_prod_df = (
    spark.read.format("csv")
    .option("header", True)
    .option("inferSchema", True)
    .load("abfss://adb-container@qcadlsbatch10.dfs.core.windows.net/raw_data/products-1gb.csv")
)

display(new_prod_df)
```

```python
new_prod_df.groupBy("category").count().display()
```

```python
(
    new_prod_df.write.format("csv")
    .option("header", True)
    .mode("overwrite")
    .save("abfss://adb-container@qcadlsbatch10.dfs.core.windows.net/processed_data/all_purpose_cluster/big_file/products")
)
```

```python
(
    new_prod_df.write.format("csv")
    .option("header", True)
    .partitionBy("category")
    .mode("overwrite")
    .save("abfss://adb-container@qcadlsbatch10.dfs.core.windows.net/processed_data/all_purpose_cluster/big_file/products")
)
```

```python
sales_df = spark.table("samples.bakehouse.sales_transactions")
display(sales_df)
```

```python
sales_df = (
    sales_df
    .withColumns({
        "year": F.year(F.col("dateTime")),
        "month": F.month(F.col("dateTime")),
        "day": F.dayofmonth(F.col("dateTime")),
    })
)
sales_df.display()
```

```python
(
    sales_df
    .write
    .format("csv")
    .option("header", True)
    .partitionBy("year", "month", "day")
    .mode("overwrite")
    .save("abfss://adb-container@qcadlsbatch10.dfs.core.windows.net/processed_data/all_purpose_cluster/sales_transactions")
)
```

```python
df = spark.read.csv("abfss://adb-container@qcadlsbatch10.dfs.core.windows.net/processed_data/all_purpose_cluster/sales_transactions", header=True)
df.display()
```
