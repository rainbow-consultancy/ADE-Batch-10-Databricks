```python
# Broadcast Join
```

```python
from pyspark.sql import functions as F
```

```python
# Disable auto broadcast join
spark.conf.set("spark.sql.autoBroadcastJoinThreshold", -1) # default is 10 MB

# Disable Adaptive Query Execution (AQE)
spark.conf.set("spark.sql.adaptive.enabled", False)
```

```python
cus_df = spark.table("samples.bakehouse.sales_customers")
trans_df = spark.table("samples.bakehouse.sales_transactions")
```

```python
display(cus_df)
```

```python
display(trans_df)
```

```python
result_df = (
    cus_df
    .join(trans_df, on=["customerID"], how="inner")
)
result_df.display()
```

```python
brod_result_df = (
    cus_df
    .join(
        F.broadcast(trans_df), 
        on=["customerID"],
        how="inner"
    )
)
brod_result_df.display()
```

```python
# Enable auto broadcast join
spark.conf.set("spark.sql.autoBroadcastJoinThreshold", 10485760) # default is 10 MB

# Enable Adaptive Query Execution (AQE)
spark.conf.set("spark.sql.adaptive.enabled", True)
```

```python
no_bro_result_df = (
    cus_df
    .join(trans_df, on=["customerID"], how="inner")
)
no_bro_result_df.display()
```

```python
brod_result_df1 = (
    cus_df
    .join(
        F.broadcast(trans_df), 
        on=["customerID"],
        how="inner"
    )
)
brod_result_df1.display()
```
