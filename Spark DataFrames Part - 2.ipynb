```python
goods_df = (
    spark.read.format("csv")
    .option("header", "true")
    # .option("inferSchema", True)
    .load("/Volumes/qc_databricks_batch_10/qc_db/quant_files/goods_classification.csv")
)
# goods_df.display()
```

```python
goods_df = (
    goods_df
    .withColumn("NZHSC_Level_2_Code_HS4", F.col("NZHSC_Level_2_Code_HS4").cast("int"))
    .withColumn("NZHSC_Level_1_Code_HS2", F.col("NZHSC_Level_1_Code_HS2").cast("int"))
)
goods_df.display()
```

```python
# latest spark functionality

goods_df = (
    goods_df
    .withColumns(
        {
            "NZHSC_Level_2_Code_HS4": F.col("NZHSC_Level_2_Code_HS4").cast("int"),
            "NZHSC_Level_1_Code_HS2": F.col("NZHSC_Level_1_Code_HS2").cast("int")
        }
    )
)

goods_df.display()
```

```python
test_df = (
    goods_df
    .select(
        F.col("NZHSC_Level_1_Code_HS2").cast("int"),
        F.col("NZHSC_Level_2_Code_HS4").cast("int"),
    )
)
test_df.display()
```

```python
test_df = goods_df
```

```python
# rename all the columns into lower case
# using for loop and withColumnRenamed 

for col in test_df.columns:
    test_df = test_df.withColumnRenamed(col, col.lower())
```

```python
test_df.display()
```

```python
# DAG - Direct Acyclic Graph - a graph with no cycles (Spark is a lazy evolution system)

1. transformations (read, filter, withColumn, select)  
2. actions (collect, show, display, write, count, first)

catalyst optimzer - spark engine -- dag -- logical plan -- optimed logical plan -- physical plan1, physical plan2, physical plan3 -- executes optimized physical plan
```

```python
old_columns = test_df.columns
new_columns = [col.lower() for col in old_columns]

test_df = (
    test_df
    .select(
        *[F.col(ocol).alias(ncol) for ncol, ocol in zip(new_columns, old_columns)]
    )
)
test_df.display()
```
