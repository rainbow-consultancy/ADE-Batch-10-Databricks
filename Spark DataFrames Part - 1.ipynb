```python
result_df = (
    cus_df
    .select(
        F.col("customerID"),
        F.col("first_name"),
        F.col("country"),
        F.col("gender")
    )
)

result_df.display()
```

```python
# filtering data from a dataframe

result_df = (
    cus_df
    .filter(F.col("gender") == 'male')
)
result_df.display()
```

```python
# filtering data from a dataframe

result_df = (
    cus_df
    .filter(
        (F.col("gender") == 'male') &
        (F.col("country") == 'USA')
    )
)
result_df.display()
```

```python
# filtering data from a dataframe

result_df = (
    cus_df
    .where(
        (F.col("gender") == 'male') &
        (F.col("country") == 'USA')
    )
)
result_df.display()
```

```python
taxi_df = spark.table("samples.nyctaxi.trips")
taxi_df.display()
```

```python
# adding a new colummn called payable_amount by multiplying with fare_amount + 10% gst

result_df = (
    taxi_df
    .withColumn(
        "payable_amount", 
        (F.col("fare_amount") * 0.10) + F.col("fare_amount"))
)

result_df.display()
```

```python
result_df = (
    taxi_df
    .withColumn(
        "payable_amount", 
        (F.col("fare_amount") * 0.10) + F.col("fare_amount"))
    .withColumn(
        "dummy",
        F.lit("True")
    )
    .withColumn(
        "dummy1",
        F.lit("False")
    )
)

result_df.display()
```

```python
result_df = (
    taxi_df
    .withColumnRenamed("tpep_pickup_datetime", "pickup_datetime")
    .withColumnRenamed("tpep_dropoff_datetime", "dropoff_datetime")
)
result_df.display()
```
