```python
from pyspark.sql import functions as F
```

```python
# Reading a table from the catalog

cus_df = spark.table("samples.bakehouse.sales_customers")
cus_df.display()
```

```python
cus_df.show()
```

```python
# Read the table using sql query

cus_df = spark.sql("select customerID, first_name, country from samples.bakehouse.sales_customers")
cus_df.display()
```

```python
result_df = (
    cus_df
    .select(
        "customerID",
        "first_name",
        "country",
        "gender"
    )
)

result_df.display()
```

```python
result_df = (
    cus_df
    .select(
        cus_df["customerID"],
        cus_df["first_name"],
        cus_df["country"],
        cus_df["gender"]
    )
)

result_df.display()
```

```python
result_df = (
    cus_df
    .select(
        cus_df.customerID,
        cus_df.first_name,
        cus_df.country,
        cus_df.gender
    )
)

result_df.display()
```

```python
result_df = (
    cus_df
    .select(
        cus_df["customerID"],
        cus_df["first_name"],
        cus_df["country"],
        cus_df["gender"]
    )
)

result_df.display()
```

```python
result_df = (
    cus_df
    .select(
        F.col("customerID"),
        F.col("first_name"),
        F.col("country"),
        F.col("gender")
    )
)

result_df.display()
```

```python
# filtering data from a dataframe

result_df = (
    cus_df
    .filter(F.col("gender") == 'male')
)
result_df.display()
```

```python
# filtering data from a dataframe

result_df = (
    cus_df
    .filter(
        (F.col("gender") == 'male') &
        (F.col("country") == 'USA')
    )
)
result_df.display()
```

```python
# filtering data from a dataframe

result_df = (
    cus_df
    .where(
        (F.col("gender") == 'male') &
        (F.col("country") == 'USA')
    )
)
result_df.display()
```
