```python
# Cache & Persist
```

```python
spark.conf.set("spark.sql.adaptive.enabled", False)
spark.conf.set("spark.databricks.io.cache.enabled", False)
```

```python
from pyspark.sql import functions as F
from pyspark import StorageLevel
```

```python
df = (
    spark.range(1, 10_000_000+1)
    .select(
        F.col("id").alias("product_id"),
        (F.col("id")%10).alias("country_id"),
        (F.col("id")*100).alias("sales")
    )
)

df.display()
```

```python
# Expensive Transaformations
filterd_df = df.filter(F.col("sales") > 50_000)


# cache the df for reuse
# filterd_df.cache().first()

filterd_df.persist(StorageLevel.DISK_ONLY)

# triggers spark action
filterd_df.count()
```

```python
first_result_df = (
    filterd_df
    .groupBy("country_id")
    .agg(F.sum("sales").alias("total_sales"))
)

first_result_df.show()
```
