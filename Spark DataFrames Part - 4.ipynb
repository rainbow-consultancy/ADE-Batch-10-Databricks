```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window
```

```python
# Read customer data

cus_df = (
    spark.table("samples.bakehouse.sales_customers")
)

trans_df = (
    spark.table("samples.bakehouse.sales_transactions")
)
```

```python
cus_df.display()
```

```python
trans_df.display()
```

```python
trans_df2 = (
    trans_df.select("transactionID", "customerID", "product", "totalPrice")
    .filter(F.col("totalPrice") > 100)
)
```

```python
cus_trans_df = (
    cus_df.select("customerID", "first_name")
    .join(
        trans_df2,
        on=["customerID"],
        how="inner"
    )
)

cus_trans_df.display()
```

this code is to demo window functions

```python
cus_trans_df = (
    cus_df
    .join(
        trans_df,
        on=["customerID"],
        how="inner"
    )
    .select("customerID", "first_name", "transactionID", "product", "totalPrice")
)

cus_trans_df.display()
```

```python
cus_trans_df = (
    cus_df
    .join(
        trans_df,
        on=["customerID"],
        how="inner"
    )
)

cus_trans_df.display()
```

```python
cus_df2 = cus_df.select(F.col("customerID").alias("cust_id").cast("string"), "first_name")
trans_df2 = trans_df.select("transactionID", F.col("customerID").alias("customer_id"), "product", "totalPrice")
```

```python
cus_df2.display()
```

```python
trans_df2.display()
```

```python
cus_trans_df = (
    cus_df2
    .join(
        trans_df2,
        on=cus_df2.cust_id == trans_df2.customer_id,
        how="inner"
    )
)

cus_trans_df.display()
```

```python
cus_trans_df = (
    cus_df2
    .join(
        trans_df2,
        on=cus_df2["cust_id"] == trans_df2["customer_id"],
        how="inner"
    )
)

cus_trans_df.display()
```

```python
# Find out the top transaction (highest totalPrice) for each customer

windowSpec = Window.partitionBy("customerID").orderBy(F.col("totalPrice").desc())

result_df = (
    trans_df
    .withColumn("rn", F.row_number().over(windowSpec))
    .filter(F.col("rn") == 1)
    .drop("rn")
)

result_df.display()
```

```python
# Find out the second top transaction (highest totalPrice) for each customer

result_df = (
    trans_df
    .withColumn("rn", F.row_number().over(windowSpec))
    .filter(F.col("rn") == 2)
    .drop("rn")
)

result_df.display()
```
