```python
from pyspark.sql import functions as F
```

```python
# read json file
cus_file_path = "/Volumes/qc_databricks_batch_10/qc_db/quant_files/customers.json"
cus_df = (
    spark.read.format("json")
    .option("multiline", "true")
    .load(cus_file_path)
)

cus_df.display()
```

```python
cus_df = (
    cus_df
    .select(
        F.col("customer_id").alias("cus_id"),
        F.col("name").alias("cus_name"),
        F.col("email").alias("email_id"),
        F.col("address.city").alias("city"),
        F.col("address.state").alias("state"),
        F.col("orders")
    )
)

cus_df.display()
```

```python
cus_df = (
    cus_df
    .withColumn("orders", F.explode(F.col("orders")))
    .select(
        F.col("cus_id"),
        F.col("cus_name"),
        F.col("email_id"),
        F.col("city"),
        F.col("state"),
        F.col("orders.order_id").alias("order_id"),
        F.col("orders.amount").alias("amount")

    )
)
cus_df.display()
```

```python
airport_file_path = "/Volumes/qc_databricks_batch_10/qc_db/quant_files/airports.json"

airport_df = (
    spark.read.format("json")
    .option("multiline", True)
    .load(airport_file_path)
)
airport_df.display()
```

```python
airport_df = airport_df.withColumn("airports", F.explode(F.col("airports")))
airport_df.display()
```

```python
airport_df = (
    airport_df
    .select(
        F.col("airports.city").alias("city"),
        F.col("airports.code").alias("code"),
        F.col("airports.country").alias("country"),
        F.col("airports.lat").alias("latitude"),
        F.col("airports.lon").alias("longitude"),
        F.col("airports.name").alias("airport_name"),
    )
)

airport_df.display()
```

```python
json_file_path = "/Volumes/qc_databricks_batch_10/qc_db/quant_files/employees_1MB.json"

emp_df = (
    spark.read.format("json")
    .option("multiline", True)
    .load(json_file_path)
)

emp_df.display()
```

```python
emp_df = (
    emp_df
    .select(
        F.col("employee.id").alias("employee_id"),
        F.col("employee.name").alias("employee_name"),
        F.col("employee.position"),
        F.col("employee.department.id").alias("department_id"),
        F.col("employee.department.name").alias("department_name"),
        F.col("employee.department.manager.id").alias("manager_id"),
        F.col("employee.department.manager.name").alias("manager_name"),
        F.col("employee.department.manager.contact.email").alias("manager_email"),
        F.col("employee.department.manager.contact.phone").alias("manager_phone"),
        F.col("employee.projects")
    )
)

emp_df.display()
```

```python
emp_df = (
    emp_df
    .withColumn("projects", F.explode(F.col("projects")))
    .select(
        F.col("employee_id"),
        F.col("employee_name"),
        F.col("position"),
        F.col("department_id"),
        F.col("department_name"),
        F.col("manager_id"),
        F.col("manager_name"),
        F.col("manager_email"),
        F.col("manager_phone"),
        F.col("projects.projectId").alias("project_id"),
        F.col("projects.projectName").alias("project_name"),
        F.col("projects.tasks").alias("project_tasks")
    )

)

emp_df.display()
```
