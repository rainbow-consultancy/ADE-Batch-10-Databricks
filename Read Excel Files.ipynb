```python
# Reading Excel Files with spark-excel
# spark-excel is a community library that allows Spark to read and write Excel files (.xls, .xlsx)
# Install via Maven coordinates: com.crealytics:spark-excel_2.12:3.5.1_0.20.4
```

```python
from pyspark.sql import functions as F
```

```python
df = (
    spark.read.format("excel")
    .option("headerRows", 1)
    .option("inferSchema", True)
    .load("/Volumes/qc_databricks_batch_10/qc_db/quant_files/Customer.xlsx")
)

df.display()
```

```python
df = (
    spark.read.format("excel")
    .option("headerRows", 1)
    .option("inferSchema", True)
    .option("sheetName", "MyWorkSheet")
    .load("/Volumes/qc_databricks_batch_10/qc_db/quant_files/Customer.xlsx")
)

df.display()
```
