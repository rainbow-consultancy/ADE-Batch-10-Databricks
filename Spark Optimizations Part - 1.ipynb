```python
# Read Products csv file from adls

prod_df = (
    spark.read.format("csv")
    .option("header", True)
    .option("inferSchema", True)
    .load("abfss://adb-container@qcadlsbatch10.dfs.core.windows.net/raw_data/products-2000000.csv")
)

prod_df.display()
```

```python
(
    prod_df.write.format("csv")
    .option("header", True)
    .mode("overwrite")
    .save("abfss://adb-container@qcadlsbatch10.dfs.core.windows.net/processed_data/all_purpose_cluster/products")
)
```

```python
# 1 GB = 1024 MB
# 1024/128 = 8
```

```python
prod_df = prod_df.repartition(3)
(
    prod_df.write.format("csv")
    .option("header", True)
    .mode("overwrite")
    .save("abfss://adb-container@qcadlsbatch10.dfs.core.windows.net/processed_data/products")
)
```
